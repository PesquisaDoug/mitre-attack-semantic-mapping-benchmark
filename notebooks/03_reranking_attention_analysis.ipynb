# Reranking And Attention Analysis

Loads Cross-Encoder reranking gain, failure cases and exploratory attention outputs. Attention weights are not interpreted as causal explanation.

```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
gain = pd.read_csv(ROOT / 'results' / 'reranking_gain.csv')
failures = pd.read_csv(ROOT / 'results' / 'strict_crossencoder_failure_cases.csv')
attention = json.loads((ROOT / 'results' / 'attention_example.json').read_text(encoding='utf-8'))
tokens = pd.read_csv(ROOT / 'results' / 'attention_token_weights.csv')
display(gain)
display(failures.head(10))
display(attention)
display(tokens.loc[~tokens['is_special']].sort_values('cls_attention', ascending=False).head(25))
```
